```python
# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier


# Step 2: Create the dataset
data = {
    "weight": [150, 170, 160, 180, 200, 190,
               120, 130, 110],
    "diameter": [7.0, 7.5, 7.2, 7.0, 7.5, 7.2,
                 3.0, 3.5, 2.8],
    "fruit": ["Apple", "Apple", "Apple",
              "Orange", "Orange", "Orange",
              "Banana", "Banana", "Banana"]
}


# Step 3: Convert the dataset into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print("Our Dataset:")
print(df)


# Step 4: Visualize the dataset
for fruit in df["fruit"].unique():
    fruit_data = df[df["fruit"] == fruit]

    plt.scatter(
        fruit_data["weight"],
        fruit_data["diameter"],
        label=fruit
    )

plt.xlabel("Weight (grams)")
plt.ylabel("Diameter (cm)")
plt.title("Fruit Classification Dataset")
plt.legend()
plt.grid(True)
plt.show()


# Step 5: Select input features
X = df[["weight", "diameter"]]

# Step 6: Select target variable
y = df["fruit"]


# Step 7: Create the model
model = DecisionTreeClassifier()


# Step 8: Train the model
model.fit(X, y)


# Step 9: Give the model a new fruit to classify
new_fruit = pd.DataFrame({
    "weight": [125],
    "diameter": [3.2]
})


# Step 10: Make a prediction
prediction = model.predict(new_fruit)


# Step 11: Display the prediction
print("Predicted fruit:", prediction[0])

```

```python

# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
import numpy as np


# Step 2: Create the dataset

np.random.seed(42)

data = {
    "weight": list(np.random.randint(140, 181, 25)) +
              list(np.random.randint(170, 221, 25)) +
              list(np.random.randint(100, 141, 25)),

    "diameter": list(np.random.uniform(6.5, 8.0, 25)) +
                list(np.random.uniform(6.0, 8.0, 25)) +
                list(np.random.uniform(2.5, 4.0, 25)),

    "fruit": ["Apple"] * 25 +
             ["Orange"] * 25 +
             ["Banana"] * 25
}


# Step 3: Convert the dataset into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print("Our Dataset:")
print(df)


# Step 4: Visualize the dataset
for fruit in df["fruit"].unique():
    fruit_data = df[df["fruit"] == fruit]

    plt.scatter(
        fruit_data["weight"],
        fruit_data["diameter"],
        label=fruit
    )

plt.xlabel("Weight (grams)")
plt.ylabel("Diameter (cm)")
plt.title("Fruit Classification Dataset")
plt.legend()
plt.grid(True)
plt.show()


# Step 5: Select input features
X = df[["weight", "diameter"]]


# Step 6: Select target variable
y = df["fruit"]


# Step 7: Create the model
model = DecisionTreeClassifier()


# Step 8: Train the model
model.fit(X, y)


# Step 9: Give the model a new fruit to classify
new_fruit = pd.DataFrame({
    "weight": [125],
    "diameter": [3.2]
})


# Step 10: Make a prediction
prediction = model.predict(new_fruit)


# Step 11: Display the prediction
print("Predicted fruit:", prediction[0])
```
